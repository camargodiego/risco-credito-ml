### Configuração de Ambiente e Imports

```python
import os
import joblib
import numpy as np
import pandas as pd
import lightgbm as lgb
import plotly.graph_objects as go

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score, roc_curve

# Ajusta o diretório de execução para a raiz do projeto
if os.getcwd().endswith("notebooks"):
    os.chdir("..")

print("✅ Ambiente e LightGBM prontos para treinamento!")
```

### Carregamento do Dataset Refinado (WoE)

```python
# Lê a base processada no Notebook 02
df_train = pd.read_parquet("data/train_pipeline_woe.parquet")

X = df_train.drop(columns=["id_cliente", "target"])
y = df_train["target"]
ids = df_train["id_cliente"]

print(f"✅ Base de treino carregada: {X.shape[0]:,} linhas e {X.shape[1]} features WoE.")
```

### Validação Cruzada Estratificada ($K$-Fold) com LightGBM

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_preds = np.zeros(len(df_train))
models = []

lgb_params = {
    "objective": "binary",
    "metric": "auc",
    "boosting_type": "gbdt",
    "learning_rate": 0.03,
    "num_leaves": 31,
    "max_depth": 5,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "random_state": 42,
    "verbose": -1,
    "n_jobs": -1
}

print("🚀 Treinando LightGBM com 5-Fold Cross Validation...\n")

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y), 1):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    model = lgb.LGBMClassifier(**lgb_params, n_estimators=1000)
    model.fit(
        X_train, y_train,
        eval_set=[(X_val, y_val)],
        callbacks=[lgb.early_stopping(stopping_rounds=50, verbose=False)]
    )
    
    oof_preds[val_idx] = model.predict_proba(X_val)[:, 1]
    models.append(model)
    
    fold_auc = roc_auc_score(y_val, oof_preds[val_idx])
    print(f"Fold {fold} - ROC-AUC: {fold_auc:.4f}")

print("\n✅ Treinamento concluído com sucesso!")
```

### Cálculo de Métricas Out-Of-Fold (ROC-AUC, Gini, KS)

```python
# Cálculo global das métricas fora da amostra (OOF)
auc_oof = roc_auc_score(y, oof_preds)
gini_oof = 2 * auc_oof - 1

# Tabela para cálculo da Estatística KS
df_ks = pd.DataFrame({"target": y, "prob": oof_preds}).sort_values(by="prob", ascending=False).reset_index(drop=True)
df_ks["cum_bads"] = (df_ks["target"] == 1).cumsum() / (df_ks["target"] == 1).sum()
df_ks["cum_goods"] = (df_ks["target"] == 0).cumsum() / (df_ks["target"] == 0).sum()
ks_oof = np.abs(df_ks["cum_bads"] - df_ks["cum_goods"]).max()

print("=" * 45)
print("     DESEMPENHO MODELO LIGHTGBM (OOF)       ")
print("=" * 45)
print(f" ROC-AUC : {auc_oof:.4f}")
print(f" Gini    : {gini_oof:.4f}")
print(f" KS      : {ks_oof * 100:.2f}%")
print("=" * 45)
```

### Visualização Dinâmica - Curva ROC Interativa (Plotly)

```python
fpr, tpr, _ = roc_curve(y, oof_preds)

fig_roc = go.Figure()
fig_roc.add_trace(go.Scatter(
    x=fpr, y=tpr, 
    mode="lines", 
    name=f"LightGBM (AUC = {auc_oof:.4f})", 
    line=dict(color="#1f77b4", width=3)
))
fig_roc.add_trace(go.Scatter(
    x=[0, 1], y=[0, 1], 
    mode="lines", 
    name="Aleatório (AUC = 0.50)", 
    line=dict(color="gray", dash="dash")
))

fig_roc.update_layout(
    title="<b>Curva ROC Out-Of-Fold (OOF)</b>",
    xaxis_title="Taxa de Falsos Positivos (1 - Especificidade)",
    yaxis_title="Taxa de Verdadeiros Positivos (Sensibilidade)",
    template="plotly_white", 
    width=750, 
    height=500,
    hovermode="x unified"
)
fig_roc.show()
```

### Visualização Dinâmica - Curva KS Interativa

```python
pop_pct = np.linspace(0, 100, len(df_ks))
max_ks_idx = np.abs(df_ks["cum_bads"] - df_ks["cum_goods"]).idxmax()
pct_pop_max = (max_ks_idx + 1) / len(df_ks) * 100

fig_ks = go.Figure()
fig_ks.add_trace(go.Scatter(
    x=pop_pct, y=df_ks["cum_bads"] * 100, 
    mode="lines", 
    name="Inadimplentes (Bads)", 
    line=dict(color="#d62728", width=3)
))
fig_ks.add_trace(go.Scatter(
    x=pop_pct, y=df_ks["cum_goods"] * 100, 
    mode="lines", 
    name="Adimplentes (Goods)", 
    line=dict(color="#1f77b4", width=3)
))
fig_ks.add_vline(
    x=pct_pop_max, 
    line_dash="dash", 
    line_color="black", 
    annotation_text=f"<b>KS Máximo = {ks_oof*100:.2f}%</b>",
    annotation_position="top left"
)

fig_ks.update_layout(
    title="<b>Curva KS (Kolmogorov-Smirnov)</b>",
    xaxis_title="Percentual Acumulado da População (%)",
    yaxis_title="Percentual Acumulado de Clientes (%)",
    template="plotly_white", 
    width=800, 
    height=500,
    hovermode="x unified"
)
fig_ks.show()
```

### Importância das Features e Exportação do Artefato do Modelo

```python
os.makedirs("src/models", exist_ok=True)

# Média de importância de variáveis entre os Folds
feature_importances = np.mean([model.feature_importances_ for model in models], axis=0)
df_importance = pd.DataFrame({
    "feature": X.columns,
    "importance": feature_importances
}).sort_values(by="importance", ascending=False)

print("--- Top 10 Features mais Importantes (LightGBM) ---")
display(df_importance.head(10))

# Seleciona o modelo com melhor desempenho AUC individual nos Folds
best_fold_idx = np.argmax([model.best_score_["valid_0"]["auc"] for model in models])
best_model = models[best_fold_idx]

# Salva o modelo treinado em disco
joblib.dump(best_model, "src/models/lightgbm_credit_model.joblib")
print(f"\n💾 Modelo treinado do Fold {best_fold_idx + 1} exportado para 'src/models/lightgbm_credit_model.joblib'!")
```
